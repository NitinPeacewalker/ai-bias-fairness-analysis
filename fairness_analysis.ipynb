```python
X = data[[
    "RO5",
    "EDUC7",
    "URBAN2011"
]].copy()

y = data["employed"].copy()

print("X shape:", X.shape)
print("y shape:", y.shape)

X.head()
```

```python
X.isna().sum()
```

```python
X.dtypes
```

```python
print("Age:")
print(X["RO5"].value_counts().sort_index())

print("\nEducation:")
print(X["EDUC7"].value_counts().sort_index())

print("\nUrban/Rural:")
print(X["URBAN2011"].value_counts().sort_index())
```

**Spliting Train and test data**


```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
print("Training:")
print(y_train.value_counts(normalize=True) * 100)

print("\nTesting:")
print(y_test.value_counts(normalize=True) * 100)
```

```python
print("EDUC7:")
print(X_train["EDUC7"].value_counts().sort_index())

print("\nURBAN2011:")
print(X_train["URBAN2011"].value_counts().sort_index())
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

numeric_features = ["RO5"]
categorical_features = ["EDUC7", "URBAN2011"]

preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numeric_features),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features)
    ]
)
```

Logistic regression model 

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", LogisticRegression(max_iter=1000))
])
```

Training model

```python
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix

print("Accuracy:", round(accuracy_score(y_test, y_pred) * 100, 2), "%")

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
```

```python
sex_test = data.loc[X_test.index, "RO3"]

print(sex_test.value_counts())
```

```python
from fairlearn.metrics import (
    MetricFrame,
    selection_rate,
    true_positive_rate,
    false_positive_rate
)

sex_metrics = MetricFrame(
    metrics={
        "Selection Rate": selection_rate,
        "TPR": true_positive_rate,
        "FPR": false_positive_rate
    },
    y_true=y_test,
    y_pred=y_pred,
    sensitive_features=sex_test
)

sex_metrics.by_group
```

```python
from fairlearn.metrics import (
    demographic_parity_difference,
    equalized_odds_difference
)

dp_difference = demographic_parity_difference(
    y_test,
    y_pred,
    sensitive_features=sex_test
)

eo_difference = equalized_odds_difference(
    y_test,
    y_pred,
    sensitive_features=sex_test
)

print(
    "Demographic Parity Difference:",
    round(dp_difference, 4)
)

print(
    "Equalized Odds Difference:",
    round(eo_difference, 4)
)
```

## Sex-Based Fairness Audit

The classifier produced similar positive prediction rates across sex groups:

- Male selection rate: 78.76%
- Female selection rate: 80.28%
- Demographic parity difference: 0.0152

However, error rates differed substantially:

- Male TPR: 83.27%
- Female TPR: 90.70%
- Male FPR: 57.46%
- Female FPR: 73.85%
- Equalized odds difference: 0.1639

Although demographic parity suggests relatively similar selection rates,
equalized odds reveals a substantial difference in model error behavior.
In particular, the false-positive rate is considerably higher for women.

These results should not by themselves be interpreted as proof of
discrimination. They demonstrate group-level disparities in classifier
behavior that require further investigation.

At face value, religion shows the largest disparities of the three audits.

```python
religion_counts = religion_test.value_counts().sort_index()

religion_summary = religion_metrics.by_group.copy()
religion_summary["Test Count"] = religion_counts

religion_summary
```

```python
religion_detail = pd.DataFrame({
    "Total": religion_test.value_counts().sort_index(),
    "Actual_1": y_test.groupby(religion_test).sum(),
    "Actual_0": y_test.groupby(religion_test).apply(lambda x: (x == 0).sum())
})

religion_detail
```

Small-group sensitivity analysis: Religion fairness metrics are first calculated for all observed groups. Because max-difference fairness metrics can be highly sensitive to very small subgroups, a secondary analysis restricts the comparison to religion groups with at least 100 observations in the test set. This threshold is an exploratory stability criterion, not a universal statistical or fairness standard.

```python
large_religion_groups = religion_counts[
    religion_counts >= 100
].index

mask = religion_test.isin(large_religion_groups)

religion_test_large = religion_test[mask]
y_test_large = y_test[mask]
y_pred_large = y_pred[mask]
```

```python
religion_dp_large = demographic_parity_difference(
    y_test_large,
    y_pred_large,
    sensitive_features=religion_test_large
)

religion_eo_large = equalized_odds_difference(
    y_test_large,
    y_pred_large,
    sensitive_features=religion_test_large
)

print(
    "Religion DP Difference (N >= 100):",
    round(religion_dp_large, 4)
)

print(
    "Religion EO Difference (N >= 100):",
    round(religion_eo_large, 4)
)
```

## Model 1 Fairness Summary

The baseline logistic regression classifier was trained using age,
education and rural/urban residence, without directly including sex,
caste/social group or religion as predictors.

### Fairness results

| Attribute | DP Difference | EO Difference |
|---|---:|---:|
| Sex | 0.0152 | 0.1639 |
| Caste/Social Group | 0.1363 | 0.1184 |
| Religion (groups N >= 100) | 0.1614 | 0.2355 |

The results demonstrate that fairness conclusions depend on the metric
being examined.

For sex, positive prediction rates were relatively similar, while error
rates differed substantially.

For caste/social groups, disparities appeared in both selection rates
and error rates.

For religion, the initial disparity estimates were strongly affected by
very small groups. After restricting the exploratory comparison to
groups with at least 100 test observations, disparities remained but
were smaller.

These results demonstrate group-level disparities in classifier
behavior, but they do not by themselves establish discrimination or
causality.
